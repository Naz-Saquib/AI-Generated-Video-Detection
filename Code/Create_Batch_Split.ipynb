```python
real_train = [f"/content/drive/MyDrive/DeepFake Detection Project/Datasets/Processed/Real/Real{i}" for i in range(0, 456)]
fake_train = [f"/content/drive/MyDrive/DeepFake Detection Project/Datasets/Processed/Fake/Fake{i}" for i in range(0, 385)]
real_validation = [f"/content/drive/MyDrive/DeepFake Detection Project/Datasets/Processed/Real/Real{i}" for i in range(456, 509)]
fake_validation = [f"/content/drive/MyDrive/DeepFake Detection Project/Datasets/Processed/Fake/Fake{i}" for i in range(385, 438)]
real_test = [f"/content/drive/MyDrive/DeepFake Detection Project/Datasets/Processed/Real/Real{i}" for i in range(509, 536)]
fake_test = [f"/content/drive/MyDrive/DeepFake Detection Project/Datasets/Processed/Fake/Fake{i}" for i in range(438, 464)]
```

```python
rt = list()
for item in real_train:
    rt.append([f'{item}main.npy',f'{item}right.npy',f'{item}vertical.npy',0])
ft = list()
for item in fake_train:
    ft.append([f'{item}main.npy',f'{item}right.npy',f'{item}vertical.npy',1])
rv = list()
for item in real_validation:
    rv.append([f'{item}main.npy',f'{item}right.npy',f'{item}vertical.npy',0])
lv =list()
for item in fake_validation:
    lv.append([f'{item}main.npy',f'{item}right.npy',f'{item}vertical.npy',1])
r_test = list()
for item in real_test:
    r_test.append([f'{item}main.npy',f'{item}right.npy',f'{item}vertical.npy',0])
l_test = list()
for item in fake_test:
    l_test.append([f'{item}main.npy',f'{item}right.npy',f'{item}vertical.npy',1])
```

```python
train_data = rt + ft
test_data = r_test + l_test
val_data = rv + lv
```

```python
import random

random.shuffle(train_data)
random.shuffle(test_data)
random.shuffle(val_data)
```
